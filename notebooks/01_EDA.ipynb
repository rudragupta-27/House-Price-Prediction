## Missing-Value Investigation Summary

The Ames Housing training dataset contains 1,460 rows and 81 columns. Missing-value analysis revealed that many missing categorical values represent structural absence rather than unknown information.

Key findings:
- Missing `PoolQC` values correspond to houses with `PoolArea = 0`.
- Missing `FireplaceQu` values correspond to houses with zero fireplaces.
- Missing garage attributes correspond to houses with zero garage capacity and area.
- Most missing basement attributes correspond to houses without basements, with a small number of exceptions where a basement exists.
- Missing `Alley`, `Fence`, and `MiscFeature` values represent the absence of those features according to the dataset's data dictionary.
- `LotFrontage` has 259 missing values. Neighborhood medians vary substantially, motivating an investigation of group-wise median imputation.
- `MasVnrType` and `MasVnrArea` require joint handling because missing type values can indicate absent veneer, an unknown type, or records where both attributes are unknown.
- `Electrical` has one missing categorical value.

### Preprocessing principles
1. Preserve the distinction between structural absence and unknown values.
2. Learn imputation statistics from training data only.
3. Fit preprocessing transformations on training data and reuse them for validation, testing, and inference.
4. Compare candidate imputation strategies using cross-validation where appropriate.
5. Keep the original raw dataset unchanged.
